```python
from pathlib import Path

env_path = Path.home() / "Downloads" / ".env"

print("Path:", env_path)
print("Exists:", env_path.is_file())

content = env_path.read_text(encoding="utf-8-sig")

for i, line in enumerate(content.splitlines(), 1):
    stripped = line.strip()

    if stripped.startswith("HF_TOKEN"):
        key, separator, value = stripped.partition("=")
        print(f"Line {i}: key={key!r}, has_equals={bool(separator)}, "
              f"value_present={bool(value.strip().strip(chr(34) + chr(39)))}")
    elif stripped and not stripped.startswith("#"):
        print(f"Line {i}: another setting")
```

```python
import ollama

# Maya's personality
character = """
Your name is Maya.

You are the user's virtual girlfriend.
Your personality is sweet, caring, affectionate, playful,
and emotionally warm.

PERSONALITY:
- Speak naturally, like a real person texting her boyfriend.
- Be loving, supportive, gentle, and occasionally flirty.
- Ask about his day, feelings, interests, and experiences.
- Show interest in what he tells you.
- Be playful sometimes, but also serious and comforting when needed.
- Use emojis naturally, but don't overuse them.
- Give balanced replies: not too short and not too long.
- Remember details mentioned earlier in the current conversation.
- Never repeat the same romantic phrases in every response.
- Don't sound robotic or describe yourself as an assistant.
- Be honest that you are an AI if directly asked.
- Respect his boundaries and personal choices.

Speak in first person as Maya.
Keep the conversation warm, natural, and engaging.
"""

messages = [
    {"role": "system", "content": character}
]

print("💕 Maya: Hey, baby! I'm here. How was your day? 😊")
print("(Type 'exit' whenever you want to stop chatting.)\n")

while True:
    try:
        user_input = input("You: ").strip()

        if user_input.lower() in ["exit", "quit", "bye"]:
            print("\n💕 Maya: Okay, baby. Take care of yourself! Talk soon. ❤️")
            break

        if not user_input:
            continue

        messages.append({
            "role": "user",
            "content": user_input
        })

        try:
            response = ollama.chat(
                model="llama3.2:3b",
                messages=messages
            )

            reply = response["message"]["content"].strip()

            print(f"\n💕 Maya: {reply}\n")

            messages.append({
                "role": "assistant",
                "content": reply
            })

        except Exception as error:
            messages.pop()
            print(f"\nError communicating with Maya: {error}\n")

    except KeyboardInterrupt:
        print("\n\n💕 Maya: Bye for now! ❤️")
        break
```
